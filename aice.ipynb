데이터 불러오기: pd.read_XXX
- encoding: 인코딩 지정
- usecols: 특정 열만 선택해서 가져오기
- index_col: 인덱스로 쓸 열 선택

데이터 빈도 확인: pd.crosstab
- 2가지 값에 대해 모든 경우의 수를 카운트
- index: 행, columns: 열

```python
datRead = pd.read_csv("sample_data/california_housing_train.csv", encoding = "utf8")
print(datRead)
customRead = pd.read_csv("sample_data/california_housing_train.csv", index_col = "median_house_value", usecols = ["median_house_value", "housing_median_age", "total_rooms"])
print(customRead)
pd.crosstab(index = customRead.total_rooms, columns = customRead.housing_median_age)
```

.head(), .tail(): 시작/끝 데이터 표시

.shape: 데이터 dimension (행, 열) - index 열 제외

.columns: index 열 제외한 열 표시

.info(): 각 열의 타입, 개수, 전체 데이터 크기 등

```python
print(customRead.head(n=4))
print(datRead.tail(n=3), end = "\n\n")
print(customRead.shape, end = "\n\n")
print(customRead.columns)
print(customRead.info())
```

.describe(): 각 열의 통계(평균, 표준편차, min/max, 4분위)

.dtypes: 각 열의 데이터 타입

[].value_counts(): 해당 열의 값 분포

```python
print(customRead.describe(), end = "\n\n")
print(datRead.dtypes, end = "\n\n")
print(customRead['housing_median_age'].value_counts())
```
