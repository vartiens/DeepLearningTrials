데이터 불러오기: pd.read_XXX
- encoding: 인코딩 지정
- usecols: 특정 열만 선택해서 가져오기
- index_col: 인덱스로 쓸 열 선택

데이터 빈도 확인: pd.crosstab
- 2가지 값에 대해 모든 경우의 수를 카운트
- index: 행, columns: 열

```python
datRead = pd.read_csv("sample_data/california_housing_train.csv", encoding = "utf8")
print(datRead)
customRead = pd.read_csv("sample_data/california_housing_train.csv", index_col = "median_house_value", usecols = ["median_house_value", "housing_median_age", "total_rooms"])
print(customRead)
pd.crosstab(index = customRead.total_rooms, columns = customRead.housing_median_age)
```

.head(), .tail(): 시작/끝 데이터 표시

.shape: 데이터 dimension (행, 열) - index 열 제외

.columns: index 열 제외한 열 표시

.info(): 각 열의 타입, 개수, 전체 데이터 크기 등

```python
print(customRead.head(n=4))
print(datRead.tail(n=3), end = "\n\n")
print(customRead.shape, end = "\n\n")
print(customRead.columns)
print(customRead.info())
```

.describe(): 각 열의 통계(평균, 표준편차, min/max, 4분위)

.dtypes: 각 열의 데이터 타입

[].value_counts(): 해당 열의 값 분포

```python
print(customRead.describe(), end = "\n\n")
print(datRead.dtypes, end = "\n\n")
print(customRead['housing_median_age'].value_counts())
```

.loc & .iloc - 행 선택

np.arange() 함수로 인덱스를 바꿀 수 있다.

이때 바꾼 인덱스 기준 -> loc

(열 선택시에도 이름 사용 가능)

그냥 0부터 시작하는 인덱스 기준 -> iloc

(열 선택시에도 인덱스 only)

```python
datRead.index = np.arange(6969, 23969)
print(datRead)
print(datRead.loc[6974, ['longitude', 'latitude']])
print(datRead.iloc[[892, 893, 894], [4, 5, 6]])
```

# 조건을 통한 데이터 선택

접근시 [] 안에 조건문을 집어넣으면 그걸 기준으로 필터링이 가능하다.

조건문은 따로 변수로 빼도 된다.

**주의**

Python은 &&, ||가 아니라 and, or을 쓴다.

그러나 pandas 사용 시에는 반드시 비트 연산자를 사용해야만 한다.


```python
filtered = datRead[datRead['latitude'] >= 38]
print(filtered)
cond = (datRead['housing_median_age'] < 25) & (datRead['population'] > 800)
fil2 = datRead[cond][['median_income', 'median_house_value']]
print(fil2)
```

열 추가하기

['새 열 이름']

기존 열로 연산해 새 열 생성 가능

.insert() 함수 사용시 위치 지정 가능

```python
filtered['newlat'] = filtered['latitude'] - 38
print(filtered)
fil3 = datRead[cond][['median_income', 'median_house_value']] # 이미 있는 열은 재삽입이 안 되므로 새로 생성
fil3.insert(2, 'houseVal_won', fil2['median_house_value'] * 1400)
print(fil3[100:1000:50])
```

열 삭제하기

.drop()

**이 함수는 DF 객체 자체를 수정하지 않는다**
따라서 함수의 return값을 저장해야 함

원본을 수정하려면 inplace = True 설정

axis = 1 -> 열 삭제

axis = 0 -> 행 삭제

```python
fil3_cut = fil3.drop('houseVal_won', axis = 1)
print(fil3_cut)
fil3_sex = fil3.drop(index = 8914, axis = 0)
print(fil3_sex[100:1000:50])
fil3_sex.drop('median_house_value', axis = 1, inplace = True)
print(fil3_sex[1077:1078])
```

열 이름 바꾸기: .rename

'기존이름': '새이름' 형태

행 단위 정렬하기: .sort_values

ascending = True이면 오름차순

(이 함수들 모두 원본 수정안함)


```python
renamed = datRead.rename(columns = {'population': 'humanCnt', 'median_income': 'nomuhyun'})
print(renamed.head(3))
sorted = datRead.sort_values('housing_median_age', ascending = True)
print(sorted.head(4))
```
