## .loc & .iloc - 행, 열 선택

**이것은 함수가 아니다!! 접근시 대괄호 사용!!**

loc[row, col] 형태로 사용

row, col에는 레이블, 인덱스, 조건 등이 들이간다.

**만약 특정 col 전체를 선택하고 싶으면 loc[:, col] 형태로 한다**

조건에 맞는 행/열 전체, 내지는 특정 셀들이 선택된다.

이렇게 선택된 값은 하나의 변수처럼 대입, 연산 등을 수행할 수 있다.

-> 그러면 모든 값들이 동시에 바뀐다.

---

np.arange() 함수로 인덱스를 바꿀 수 있다.

이때 바꾼 인덱스 기준 -> loc

(열 선택시에도 이름 사용 가능)

그냥 0부터 시작하는 인덱스 기준 -> iloc

(열 선택시에도 인덱스 only)

```python
datRead.index = np.arange(6969, 23969)
print(datRead)
print(datRead.loc[6974, ['longitude', 'latitude']])
print(datRead.iloc[[892, 893, 894], [4, 5, 6]])
```

# 조건을 통한 데이터 선택

접근시 [] 안에 조건문을 집어넣으면 그걸 기준으로 필터링이 가능하다.

조건문은 따로 변수로 빼도 된다.

**주의**

Python은 &&, ||가 아니라 and, or을 쓴다.

그러나 pandas 사용 시에는 반드시 비트 연산자를 사용해야만 한다.


```python
filtered = datRead[datRead['latitude'] >= 38]
print(filtered)
cond = (datRead['housing_median_age'] < 25) & (datRead['population'] > 800)
fil2 = datRead[cond][['median_income', 'median_house_value']]
print(fil2)
```

열 추가하기

['새 열 이름']

기존 열로 연산해 새 열 생성 가능

.insert() 함수 사용시 위치 지정 가능

```python
filtered['newlat'] = filtered['latitude'] - 38
print(filtered)
fil3 = datRead[cond][['median_income', 'median_house_value']] # 이미 있는 열은 재삽입이 안 되므로 새로 생성
fil3.insert(2, 'houseVal_won', fil2['median_house_value'] * 1400)
print(fil3[100:1000:50])
```

열 삭제하기

.drop()

**이 함수는 DF 객체 자체를 수정하지 않는다**
따라서 함수의 return값을 저장해야 함

원본을 수정하려면 inplace = True 설정

axis = 1 -> 열 삭제

axis = 0 -> 행 삭제

```python
fil3_cut = fil3.drop('houseVal_won', axis = 1)
print(fil3_cut)
fil3_sex = fil3.drop(index = 8914, axis = 0)
print(fil3_sex[100:1000:50])
fil3_sex.drop('median_house_value', axis = 1, inplace = True)
print(fil3_sex[1077:1078])
```

데이터의 그룹화

한 열의 데이터를 기준으로 해서 전체 데이터를 묶을 수 있다.

Split -> Apply -> Combine

```python
groupDat = datRead.groupby('housing_median_age')
print(groupDat.groups) # 객체의 모든 값을 단순 표시
```

groupby 객체에서 함수를 적용해 통계를 구할 수 있다.

```python
print(groupDat.count()[['longitude']], end='\na==============a\n') # 각 값에 해당하는 데이터 수(모든 열)
print(groupDat.size(), end='\na==============a\n') # 해당 열 데이터의 각 값당 개수
print(groupDat.max(), end='\na==============a\n') # 각 그룹의 각 열별 최댓값
print(groupDat.min(), end='\na==============a\n') # 최솟값
print(groupDat.mean(), end='\na==============a\n') # 평균, 이외에도 std, var 등 존재
```

groupby는 두 개 이상의 열에 대해서도 할 수 있다.

```python
datR_newV = pd.read_csv("sample_data/california_housing_train.csv")
datR_newV['lat'] = datR_newV['latitude'].astype(int)
datR_newV = datR_newV.drop(['latitude'], axis = 1)
print(datR_newV.groupby(['housing_median_age', 'lat']).mean())
```

loc을 사용하면 원하는 인덱스의 데이터만 꺼낼 수 있다.

이때 index가 다차원으로 돼 있으면 Tuple을 사용해야만 한다.

```python
datRVgrp = datR_newV.groupby(['housing_median_age', 'lat'])
print(datRVgrp.mean().loc[[(10.0, 38), (20.0, 40)]])
```

index 설정 시에도 이러한 다차원 group을 적용 가능하다.

set_index를 해놓으면 groupby 사용 시 level을 지정할 수 있다.

이 level은 먼저 세팅된 순서대로 0, 1, 2, ...로 진행한다.

값을 어떻게 주냐에 따라 순서를 바꿀 수도 있다.

이때는 반드시 index 설정 **먼저 해두고** groupby 해줘야 한다.

```python
indexedRV = datR_newV.set_index(['housing_median_age', 'lat'])
print(indexedRV.groupby(level = [1]).mean())
print(indexedRV.groupby(level = [0, 1]).std())
print(indexedRV.groupby(level = [0, 1]).aggregate([np.min, np.max])) # np.min, max는 뭐에 쓰는 거지?
```

# Pivot Table

데이터 값들을 행/열 index로 사용하는 것이 가능하다.

위에서 set_index 한 것을 2차원 표로 펼치는 느낌

DataFrame의 .pivot/.pivot_table 함수로 생성 가능

둘의 차이: index 겹치는 경우 에러(pivot), 평균값(pivot_table)

겹치는 값에서는 aggfunc로 평균값 대신 다른거 쓸수도 있음(np.max, np.sum 등등)

```python
# datR_newV.pivot(columns = 'housing_median_age', index = 'lat', values = 'population')
# ValueError: Index contains duplicate entries, cannot reshape
datR_newV.pivot_table(columns = 'housing_median_age', index = 'lat', values = 'population', aggfunc = np.std)
```

unstack: 지정된 index를 열로 이동

stack: 다시 index로 이동

```python
print(indexedRV.unstack(0))
print(indexedRV.unstack(1))
```

데이터 병합: pd.concat

여기도 파라미터 모두 list로 묶어서 줘야함

ignore_index: 인덱스 강제 재배치 여부

axis: 어느 방향으로 붙일지 (0: 행(아래로), 1: 열(오른쪽으로))

join: 열이 안 맞을 때 처리 방법 (inner: 교집합만 남김, outer: 전부 다 남김)

verify_integrity: index 중복시 에러 표출

```python
t1 = pd.DataFrame({'col1': ['ar', 'sg', 'hg', 'mg', 'sr'], 'col2': [25, 50, 15, 40, 35]}, index = [0, 1, 2, 3, 4])
t2 = pd.DataFrame({'col1': ['ar', 'sg', 'mt', 'smg', 'rl'], 'col2': [20, 45, 55, 15, 40]}, index = [2, 3, 4, 5, 6])
print(pd.concat([t1, t2], axis = 1))
print(pd.concat([t1, t2], ignore_index = True))
print()
t3 = pd.DataFrame({'col2': ['ar', 'sg', 'mt', 'sex'], 'col3': [30, 55, 65, 69], 'col4': [16, 25, 36, 49]}, index = [2, 3, 4, 5])
print(pd.concat([t3, t2], join = 'outer'))
print(pd.concat([t3, t2], join = 'inner'))
```

merge & join - 특정 값을 기준으로 병합

on: 기준으로 쓸 열을 지정

how: 병합 방법 (inner, outer, left, right)

    inner, outer: 위와 같음, inner가 기본

    left, right: 왼쪽/오른쪽을 기준으로 병합 (그쪽에 없는 행은 다 무시)

left_index, right_index: True로 하면 설정된 인덱스를 활용

(사용시 on 제거필요)

```python
print(pd.merge(t1, t2, on = 'col1', how = 'outer'))
print(pd.merge(t2, t1, on = 'col1', how = 'left'))
tx1 = t1.set_index('col1')
tx3 = t3.set_index('col2')
print(pd.merge(tx1, tx3, how = 'right', left_index = True, right_index = True))
```
